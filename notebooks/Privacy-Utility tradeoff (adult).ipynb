```python
%load_ext autoreload
%autoreload 2
%matplotlib inline

from warnings import filterwarnings
filterwarnings('ignore')

import sys
sys.path.append('../')

from utils.analyse_results import *
from utils.tradeoff_plot import plot_privacy_utility_tradeoff
```

```python
linkage_results   = load_results_linkage('../tests/linkage/')
inference_results = load_results_inference('../tests/inference/', '../data/adult')
_, utility_results = load_results_utility('../tests/utility/')

# Filter to adult dataset only
linkage_results   = linkage_results[linkage_results['Dataset'] == 'adult']
inference_results = inference_results[inference_results['Dataset'] == 'adult']
utility_results   = utility_results[utility_results['Dataset'] == 'adult']

print('Linkage models:', sorted(linkage_results['TargetModel'].unique()))
print('Inference attrs:', sorted(inference_results['SensitiveAttribute'].unique()))
print('Utility LabelVar:', sorted(utility_results['LabelVar'].unique()))
```

# Privacy-Utility Tradeoff (Adult)

```python
# SensitiveAttribute = HoursPerWeek
fig = plot_privacy_utility_tradeoff(
    linkage_results, inference_results, utility_results,
    featureset_filter='Correlations',
    sensitive_attr_filter='HoursPerWeek',
    label_var_filter='label',
    pred_model_filter='RandomForestClassifier',
)
fig.savefig('tradeoff_adult_hours.pdf', bbox_inches='tight')
```
